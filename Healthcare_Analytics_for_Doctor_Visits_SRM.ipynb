## **6.1 Univariate Visulization**




# 1. Age Distribution - *Histogram*
>This visualization shows the distribution of patients across different age groups. It helps us identify the most common age groups and understand the overall age pattern of patients.

```python
plt.figure(figsize=(10, 6))

# Create Histogram
sns.histplot(
    data=df,
    x='age',
    bins=10,
)

plt.title('Age Distribution of Patients', fontsize=16, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel("Number of Patiuents", fontsize=12)
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

# 2. Gender Distribution of Patients - *Pie Chart*

> This visualization shows the proportion of male and female patients, helping us understand the overall gender distribution in the dataset.

```python
# Count each gender
gender_count = df['gender'].value_counts()

# Create chart
plt.figure(figsize=(7, 7))

# Plot pie chart
wedges, texts, autotexts = plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)

# Set percentage color
for text in autotexts:
    text.set_color('white')
    text.set_fontweight('bold')

plt.axis('equal')
plt.title('Gender Distribution of Patients', fontsize=16, fontweight='bold')

# Display chart
plt.show()
```

# 3. Distribution of Patients Visits - *Count Plot*

> This visualization identifies the most common visit count and shows how the number of patients varies across differrnt viusit counts.

```python
plt.figure(figsize=(10, 6))

# Create count plot
sns.countplot(
    data=df,
    x='visits',
    color='#2A9D8F'
)

plt.title(
    'Distribution of Patient Visits',
    fontsize=16,
    fontweight='bold',
    color='#264653',
    pad=15
)


plt.xlabel('Number of Visits', fontsize=12)
plt.ylabel('Number of Patients', fontsize=12)

plt.grid(axis='y', alpha=0.2, linestyle='--')

plt.tight_layout()

plt.show()
```

# 4. Patient Distribution by Illness - *Horizontal Bar Chart*

> This Horizontal Bar Chart compares the number of patients across different illness categories. It helps identify which illness categories have the highest and lowest patient counts.

```python
illness_count = df['illness'].value_counts().sort_values()

plt.figure(figsize=(9, 5))

# Create horizontal bars
ax = illness_count.plot(
    kind='barh',
    color='#457B9D',
    edgecolor='white'
)

plt.title(
    'Patient Distribution by Illness',
    fontsize=16,
    fontweight='bold',
    color='#264653',
    pad=15
)

plt.xlabel('Number of Patients', fontsize=12)
plt.ylabel('Illness Category', fontsize=12)

for container in ax.containers:
    ax.bar_label(container, padding=4, fontsize=10)

ax.grid(axis='x', alpha=0.2, linestyle='--')
ax.set_axisbelow(True)

plt.tight_layout()

plt.show()
```

# 5. Patient Distribution by Health Status - *Donut Chart*

> This Donut Chart shows the percentage distribution of patients across different reported health status categories. It helps identify the most common health status category among patients.

```python
health_counts = df['health'].value_counts().sort_index()

# Select top 5 categories
top_health = health_counts.nlargest(5)
top_health['Other'] = health_counts.drop(top_health.index).sum()

# Create donut chart
plt.figure(figsize=(9, 6))
wedges, texts, autotexts = plt.pie(
    top_health.values,
    labels=None,
    autopct='%1.1f%%',
    startangle=90,
    counterclock=False,
    pctdistance=0.78,
    colors=sns.color_palette('Set2', len(top_health)),
    wedgeprops={'width': 0.38, 'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 10}
)

# Add center text
plt.text(0, 0.08, 'HEALTH', ha='center', va='center',
         fontsize=13, fontweight='bold', color='#264653')
plt.text(0, -0.12, 'STATUS', ha='center', va='center',
         fontsize=11, color='#555555')

# Add title
plt.title('Patient Distribution by Health Status',
          fontsize=15, fontweight='bold', pad=18)

# Add legend
plt.legend(
    wedges,
    [f'{label} (n={count:,})' for label, count in top_health.items()],
    title='Health Category',
    loc='center left',
    bbox_to_anchor=(1.0, 0.5),
    frameon=False
)

plt.axis('equal')
plt.tight_layout()
plt.show()
```

## **6.2 Multivariate Visulization**


## 6. Average Patient Visits by Gender - *Bar Plot*

> This chart compares average patient visits by gender and helps us identify differences in the average number of visits between the two groups.

```python
plt.figure(figsize=(9, 5))

# Create Bar plot
ax = sns.barplot(
    data=df,
    x='gender',
    y='visits',
    color='#2A9D8F',
    errorbar=None
)

plt.title(
    'Average Patient Visits by Gender',
    fontsize=16,
    fontweight='bold',
    color='#264653',
    pad=15
)

plt.xlabel('Gender', fontsize=11, fontweight='bold')
plt.ylabel('Average Number of Visits', fontsize=11, fontweight='bold')

ax.tick_params(axis='x', labelsize=14)

# Add exact average values
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.2f',
        fontsize=12,
        fontweight='bold',
        padding=6
    )

# Add space above bars
ax.margins(y=0.15)


ax.grid(axis='y', alpha=0.4, linestyle='--')
ax.set_axisbelow(True)


plt.tight_layout()

plt.show()
```

# 7. Illness Score vs. Average Doctor Visits - *Line Chart*

> This Line Chart shows the relationship between illness score and average doctor visits. It helps identify how the average number of doctor visits varies across different illness scores.

```python
# Calculate illness analysis
illness_analysis = (
    df.groupby("illness")["visits"]
      .agg(["count", "mean"])
      .reset_index()
)

# Create Line Chart
sns.lineplot(
    data=illness_analysis,
    x="illness",
    y="mean",
    marker="o",
    color="#2A9D8F"
)

plt.title("Illness Score vs. Average Doctor Visits")
plt.xlabel("Illness score")
plt.ylabel("Mean recorded visits")
plt.show()
```

# 8. Healthcare Variables Correlation - *Heatmap*

> This Heatmap displays the correlation between numerical healthcare variables. Values closer to +1 indicate a positive relationship, values closer to −1 indicate a negative relationship, and values near 0 indicate a weak linear relationship.

```python
# Select numerical columns
numeric_df = df.select_dtypes(include='number').drop(
    columns=['Unnamed: 0'], errors='ignore'
)

# Calculate correlation
correlation = numeric_df.corr()

# Create Heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title(
    'Healthcare Variables Correlation Heatmap',
    fontsize=15,
    fontweight='bold',
    color='#264653',
    pad=15
)

plt.tight_layout()
plt.show()
```

# 9. Average Reduced Activity Days by Chronic Condition and Gender - *Grouped Bar Chart*

> This chart compares the average number of days with reduced activity due to health issues, categorized by the presence of a chronic condition ('nchronic') and gender. It helps in understanding the differential impact of chronic conditions on activity levels across genders.

```python
plt.figure(figsize=(10, 6))

# Create bar plot
sns.barplot(
    data=df,
    x='nchronic',
    y='reduced',
    hue='gender',
    errorbar=None,
    palette='Set2'
)

plt.title(
    'Average Reduced Activity Days by Chronic Condition and Gender',
    fontsize=15,
    fontweight='bold',
    color='#264653',
    pad=15
)

plt.xlabel(
    'Number of Chronic Conditions',
    fontsize=12,
    fontweight='bold',
    labelpad=8
)

plt.ylabel(
    'Average Reduced Activity Days',
    fontsize=12,
    fontweight='bold',
    labelpad=8
)

# Style legend and grid
plt.legend(title='Gender', loc='upper right')
plt.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

# 10. Distribution of Doctor Visits by Illness Score - *Box Plot*

> This Box Plot shows the distribution of doctor visits across different illness scores. It helps compare the median, variability, and potential outliers in doctor visits for each illness score.

```python
# Create Box Plot
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x='illness',
    y='visits',
    color='#76B7B2'
)

# Style title
plt.title(
    'Distribution of Doctor Visits by Illness Score',
    fontsize=15,
    fontweight='bold',
    color='#264653',
    pad=15
)

# Style axis labels
plt.xlabel('Illness Score', fontsize=12, labelpad=8)
plt.ylabel('Number of Doctor Visits', fontsize=12, labelpad=8)

plt.tight_layout()
plt.show()
```

# 11. Age vs. Doctor Visits by Gender - *Scatter Plot*

> This Scatter Plot shows the relationship between patient age and doctor visits, using different colors for gender categories. It helps identify patterns, clusters, and variations in visit frequency across ages and genders.

```python
plt.figure(figsize=(10, 6))

# Create scatter plot
sns.scatterplot(
    data=df,
    x='age',
    y='visits',
    hue='gender',
    palette='deep',
    alpha=0.6
)

plt.title(
    'Age vs. Doctor Visits by Gender',
    fontsize=15,
    fontweight='bold',
    color='#264653',
    pad=15
)

plt.xlabel('Age', fontsize=12, labelpad=8)
plt.ylabel('Number of Doctor Visits', fontsize=12, labelpad=8)

plt.legend(title='Gender')
plt.grid(alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```
